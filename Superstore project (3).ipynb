## **BAR CHART**
## **SALES BY CATEGORY**

```python
category_sales = df.groupby("Category")["Sales"].sum()
colors=["red","blue","green"]

category_sales.plot(kind="bar",figsize=(6,4),color=colors)
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

## **INSIGHT**

```python
The Technology category generated the highest sales , followed by furniture.office supplies recorded the lowest sales among the three categories.
```

## **HORIZONTAL BARCHART**
## ** TOP 10 PRODUCT BY SALES**

```python
top_product = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)
colors=["red","yellow","blue","purple","pink","orange","black","cyan","magenta","gold"]
top_product.plot(kind="barh",figsize=(8,5),color=colors)
plt.title("Top 10 Product by Sales")
plt.xlabel("Sales")
plt.ylabel("Product Name")
plt.show()
```

## **INSIGHT**

```python
Canon image class 2200 advanved copier was highest - selling product ,indicating high customer demand.
```

## **PIE CHART**
## **SALES BY SEGMENT**

```python
segment_sales = df.groupby("Segment")["Sales"].sum()
colors = ["yellow","red","cyan"]
segment_sales.plot(kind="pie",autopct="%1.1f%%",labels=segment_sales.index,colors=colors,shadow=True,radius=1,startangle=120,textprops={"fontsize":12}
    ,wedgeprops={"linewidth":1,"edgecolor":"black"},pctdistance=0.4,labeldistance=1.1,rotatelabels=False)
plt.title("SALES BY SEGMENT")
plt.ylabel("")
plt.show()

                                                                                                                                                            
                                                                                                                                                                    
```

## **INSIGHT**

```python
The consumer segment contributed the highest share of sales (about 50.6%), followed  by corporate (30.7%) and home office(18.7%).
```

## **SCATTER CHART**
## **PROFIT BY REGION**

```python
region_profit =df.groupby("Region")["Profit"].sum().reset_index() 
colors =["red","yellow","green","blue"]

plt.scatter(region_profit["Region"],region_profit["Profit"],cmap="virdis",s=150,c=colors,marker="*"),
plt.colorbar()
plt.title("Profit by Region") 
plt.xlabel("Region")
plt.ylabel("Profit") 
plt.grid() 
plt.show()
```

## **INSIGHT**

```python
The west region generated the highest profit, while the south region recorded the lowest profit.
the company can focus on improving profitability in lower - performing regions.
```

## **MONTHLY SALES TREND**
## **LINE CHART**

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
monthly_sales = df.groupby(df["Order Date"].dt.month)["Sales"].sum()
plt.figure(figsize=(12, 6))
monthly_sales.plot(kind="line",marker="o",linestyle="--",markerfacecolor="red",markeredgecolor="black")
for  x, y in zip(monthly_sales.index,monthly_sales.values):
    plt.annotate(f"{y/1000:.1f}k",(x,y),xytext=(0,2),textcoords="offset points",ha="center")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.legend(loc="lower right")
plt.show()
```

## **INSIGHT**

```python
Monthly sales show an overall upward trend with some fluctuations. After a few moderate months,
sales increased significantly in the later months, reaching a peak of approximately 352.5K, 
indicating stronger business performance towards the end of the period.
```

## **PIE CHART**
## **SALES BY CATEGORY**


```python
categoory_sales=df.groupby("Category")["Sales"].sum()
plt.figure(figsize=(6,6))
colors=["gold","purple","red"]
plt.pie(categoory_sales,labels=categoory_sales.index,autopct="%1.1f%%",startangle=90,colors=colors,shadow=True)
plt.title("Sales by category")
plt.ylabel("")
plt.show()
```

## **AREA CHART**
## **MONTHLY SALES TREND**

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
monthly_sales = df.groupby(df["Order Date"].dt.month)["Sales"].sum()
monthly_sales.plot(kind="area",figsize=(12,6),color="b")
plt.plot(monthly_sales.index,monthly_sales.values,color="black",marker="o",linewidth=2,mfc="r")
for  x, y in zip(monthly_sales.index,monthly_sales.values):
    plt.annotate(f"{y/1000:.1f}k",(x,y),xytext=(0,2),textcoords="offset points",ha="center")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

## **INSIGHT**

```python
Monthly sales showed an overall upward trend with some fluctuations. Sales reached the highest value (352.5K) in the later months,
indicating improved business performance.
```
